2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
常見的 Python 執行環境使用 IEEE 754，雙精度浮點數，IEEE 754 常見的預設舍入方式是 round to nearest, ties to even，中文常稱：

四捨六入五成雙

概念為：

小於一半 → 捨去
大於一半 → 進位
剛好一半 → 選擇最低有效位為偶數的結果

這樣可以降低大量運算時，永遠向同一方向進位造成的系統性偏差。搜尋資料對 sys.float_info.rounds = 1 的說明也是「round to nearest」，通常對應 IEEE 754 的最近偶數舍入。
